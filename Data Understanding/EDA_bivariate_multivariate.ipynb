```python
import pandas as pd
import seaborn as sns
```

```python
df=sns.load_dataset('tips')
df.head()
```

```python
titanic=pd.read_csv('Titanic-Dataset.csv')
titanic.head()
```

```python
flights=sns.load_dataset('flights')
flights.head()
```

```python
iris=sns.load_dataset('iris')
iris.head()
```

### Scatterplot

```python
sns.scatterplot(data=df,x='total_bill',y='tip',hue='sex',style='smoker')
```

## Bar Plot

```python
sns.barplot(data=titanic,x='Pclass',y='Fare',hue='Sex')
```

### Box Plot

```python
sns.boxplot(data=titanic,x='Sex',y='Age',hue='Survived')
```

### Dist Plot

```python
sns.distplot(titanic[titanic['Survived']==0]['Age'],hist=False)
sns.distplot(titanic[titanic['Survived']==1]['Age'],hist=False)
```

### Heat Map

```python
titanic.head(2)
```

```python
sns.heatmap(pd.crosstab(titanic['Pclass'],titanic['Survived']))
```

```python
(titanic.groupby('Pclass')['Survived'].mean() * 100).plot(kind='bar')
```

### Cluster Map

```python
sns.clustermap(pd.crosstab(titanic['SibSp'],titanic['Survived']))
```

### Pair plot

```python
iris.head()
```

```python
sns.pairplot(iris,hue='species')
```

### Line Plot

```python
flights.head()
```

```python
new = flights.groupby('year')['passengers'].sum().reset_index()
```

```python
sns.lineplot(data=new,x='year',y='passengers')
```

```python
sns.heatmap(flights.pivot_table(values='passengers',index='month',columns='year'))
```

```python
sns.clustermap(flights.pivot_table(values='passengers',index='month',columns='year'))
```
