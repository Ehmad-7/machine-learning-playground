```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('Titanic-Dataset.csv')
```

```python
df.head()
```

## Categorical Data

### Countplot

```python
sns.countplot(x="Survived",data=df,palette='pastel',hue="Survived")
df['Survived'].value_counts()
```

### PieChart

```python
df['Pclass'].value_counts().plot(kind='pie',autopct='%.2f%%',colors=['lightblue','lightgreen','lightpink'])
```

## Numerical Data

### Histogram

```python
plt.hist(df['Age'],bins=10)
```

### Distplot

```python
sns.displot(df['Age'],kde=True)
```

### Boxplot

```python
sns.boxplot(x="Fare", data=df, width=0.5)
plt.xlim(0, 80)
plt.show()
```
